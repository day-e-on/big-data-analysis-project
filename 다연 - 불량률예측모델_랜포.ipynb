### 모델 학습

```python
import pandas as pd
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report

df_raw = df_raw.drop(columns=['No_Die', 'Ox_Chamber', 'Lot_Num', 'Wafer_Num',
                              'Datetime', 'photo_soft_Chamber','lithography_Chamber','Etching_Chamber','Chamber_Num',
                              'Target', 'Error_message', 'Path'], errors='ignore')

# 불량률 예측 모델 학습
X = df_raw.drop(columns=['Target_error'])
y = df_raw['Target_error']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=4321)
```

```python
rf_model = RandomForestClassifier(random_state=12,n_estimators = 100,max_depth = 5)
rf_model.fit(X_train, y_train)
y_pred = rf_model.predict(X_train)
print("Classification Report:\n", classification_report(y_train, y_pred))
# Evaluate on the test set
y_pred = rf_model.predict(X_test)
print("Classification Report:\n", classification_report(y_test, y_pred))
```

```python
rf_model = DecisionTreeClassifier(random_state=4321)
rf_model.fit(X_train, y_train)
# Evaluate on the test set
y_pred = rf_model.predict(X_test)
print("Classification Report:\n", classification_report(y_test, y_pred))
```
