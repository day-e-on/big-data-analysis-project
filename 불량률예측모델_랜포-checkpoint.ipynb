```python
df = pd.read_csv('')
```

# 필요없는 설명변수 드랍

```python
df_cleaned = df.drop(['No_Die', 'Lot_Num', 'Wafer_Num', 'Datetime', 'Error_message'], axis=1)
```

```python
from sklearn.model_selection import train_test_split, cross_val_score
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, confusion_matrix

# 1. 데이터 준비 - 상관관계 높은 변수 제거
# 상관관계가 높은 변수들을 시각적으로 확인한 후, 대표 변수 선택을 위한 절차 필요
# 여기서는 예시로 상관관계 높은 변수들 중 일부를 임의로 제거
# 과적합 방지 목적으로 일단 상관관계 확인 없이 예시로 변수를 일부 삭제하여 실행
# X, y 분리
X = df_cleaned.drop('Target_error', axis=1)
y = df_cleaned['Target_error']

# 2. 데이터 스케일링 (정규화)
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# 3. 차원 축소 - PCA
pca = PCA(n_components=10)  # 예시로 10개 주요 성분 사용
X_pca = pca.fit_transform(X_scaled)

# 데이터 분할
X_train, X_test, y_train, y_test = train_test_split(X_pca, y, test_size=0.2, random_state=3000)

# 모델 초기화
models = {
    'Logistic Regression': LogisticRegression(max_iter=1000),
    'Decision Tree': DecisionTreeClassifier(),
    'Random Forest': RandomForestClassifier(),
    'Gradient Boosting': GradientBoostingClassifier()
}

# 각 모델별 성능 평가
results = []
for model_name, model in models.items():
    # 교차 검증으로 일반화 성능 평가
    cv_scores = cross_val_score(model, X_train, y_train, cv=5, scoring='f1')
    model.fit(X_train, y_train)
    y_pred = model.predict(X_test)
    
    # 성능 지표 계산
    accuracy = accuracy_score(y_test, y_pred)
    precision = precision_score(y_test, y_pred)
    recall = recall_score(y_test, y_pred)
    f1 = f1_score(y_test, y_pred)
    cm = confusion_matrix(y_test, y_pred)
    
    # 결과 저장
    results.append({
        'Model': model_name,
        'CV F1 Score (Mean)': cv_scores.mean(),
        'Test Accuracy': accuracy,
        'Test Precision': precision,
        'Test Recall': recall,
        'Test F1 Score': f1,
        'Confusion Matrix': cm
    })

# 각 모델별 성능 평가 후 결과 출력
for result in results:
    print(f"모델: {result['Model']}")
    print(f"교차 검증 F1 Score 평균: {result['CV F1 Score (Mean)']:.4f}")
    print(f"테스트 정확도: {result['Test Accuracy']:.4f}")
    print(f"테스트 정밀도: {result['Test Precision']:.4f}")
    print(f"테스트 재현율: {result['Test Recall']:.4f}")
    print(f"테스트 F1 Score: {result['Test F1 Score']:.4f}")
    print("오차 행렬:")
    print(result['Confusion Matrix'])
    print("\n" + "="*50 + "\n")
```
